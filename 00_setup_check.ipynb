```python
import numpy as np, lightgbm as lgb, shap

rng = np.random.default_rng(0)
X = rng.random((300, 4))
y = (X[:, 0] + rng.normal(0, 0.1, 300) > 0.5).astype(int)

model = lgb.LGBMClassifier(n_estimators=20, verbose=-1).fit(X, y)
values = shap.TreeExplainer(model).shap_values(X[:5])
print("shap values shape:", np.array(values).shape)
```
